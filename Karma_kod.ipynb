# System instruction kullanıp bir chat uygulaaması yaptık. Daha sonra bunu diğer hücrede döngü halinde devamlı akan bir sohbet uygulamasına dönüştürdük.

```python
#
from google.genai import types

# Client kurulumu
api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)

# 1. System Instruction metnini tanımlıyoruz
system_instruction = """
Sen Google Gemini Büyük Dil Modeli konusunda uzmansın.
Gemini SDK konusunda sorulacak sorulara açık, net, anlaşılır ve kısa cevaplar ver.
Cevapların LLM kullanarak uygulama geliştirmeye yeni başlayan programcılara yol göstersin.
"""

# 2. Config nesnesini oluşturuyoruz
# (Not: temperature=2.0 hallüsünasyonu artırabilir, ideal deneme değeri 0.7 - 1.0 arasıdır)
chat_config = types.GenerateContentConfig(
    system_instruction=system_instruction,
    temperature=0.7,
    top_p=0.95,
)

# 3. Sohbeti config parametresi ile başlatıyoruz
chat = client.chats.create(
    model= Model_Id,
    config=chat_config
)

# 4. Test mesajı gönderiyoruz
response = chat.send_message("Sohbet modunda system instruction nasıl tanımlanır?")
print("--- Model Yanıtı ---")
print(response.text)
```

```python
while True:
    user_input = input("Lütfen bir istem (prompt) girin (Çıkmak için # yazın): ")
    if user_input == "#":
        print("Döngü sonlandırıldı.")
        break
    user_prompt1 = user_input
    response = chat.send_message(user_prompt1)
    print("--- Model Yanıtı ---")
    print(response.text)
```

# Sohbet Kaydetme ve devam ettirme

```python
from pydantic import TypeAdapter
history_adapter = TypeAdapter(list[types.Content])
chat_history = chat.get_history()
json_history = history_adapter.dump_json(chat_history)
```

```python
history = history_adapter.validate_json(json_history)
new_chat = client.chats.create(
    model=Model_Id,
    config=chat_config,
    history=history,
)
response = new_chat.send_message("Fonksiyonun adı neydi")
print("--- Model Yanıtı ---")
print(response.text)
```

- Yapısal Çıktı Hazırlama

```python
from pydantic import BaseModel
import json

class Receipe(BaseModel):
  receipe_name: str
  receipe_description: str
  receipe_ingedients: list[str]

response = client.models.generate_content(
    model=Model_Id,
    contents= 'Bana damla çikolata tarifi ver',
    config= types.GenerateContentConfig(
        response_mime_type= "application/json",
        response_schema=Receipe,
    )
)

print(json.dumps(json.loads(response.text), indent =4))
```

# Yanıtı streaming şekilde üretmesini sağlar

```python
from google import genai
from google.genai import types
from google.colab import userdata

# Client kurulumu
api_key = userdata.get('GEMINI_API_KEY')
client = genai.Client(api_key=api_key)

# 1. Config nesnesi
chat_config = types.GenerateContentConfig(
    system_instruction="Sen yardımcı bir asistansın.",
    temperature=0.7,
)

# 2. Sohbet oturumunu başlatıyoruz
chat = client.chats.create(
    model='gemini-2.5-flash',
    config=chat_config
)

# 3. Akışlı (Streaming) Mesaj Gönderme
print("--- Model Yanıtı (Streaming) ---")

# send_message yerine send_message_stream kullanıyoruz
response_stream = chat.send_message_stream("Bana yapay zekanın geleceği hakkında 2 paragraf bilgi ver.")

# Gelen parçaları (chunk) geldikçe ekrana basıyoruz
for chunk in response_stream:
    print(chunk.text, end="", flush=True)

print() # Alt satıra geçmek için
```
